```python
import matplotlib.pyplot as plt
import numpy as np

# ==============================================================================
# 1. POST-FEC BER vs. SNR (WATERFALL CURVE)
# ==============================================================================

snrs = np.arange(0, 13)

# 0.0 values are replaced with 1e-6 so they render correctly on a log scale plot
ber_classical = [2.851e-01, 2.713e-01, 2.581e-01, 2.446e-01, 2.311e-01, 2.052e-01, 1.594e-01, 1.355e-01, 1.210e-01, 1.100e-01, 1.009e-01, 9.285e-02, 8.593e-02]
ber_ours = [2.674e-01, 2.508e-01, 2.343e-01, 2.171e-01, 1.968e-01, 1.533e-01, 3.765e-02, 8.472e-03, 6.175e-03, 5.601e-03, 5.631e-03, 5.623e-03, 5.513e-03]
ber_deeprx = [3.098e-01, 2.827e-01, 2.595e-01, 2.380e-01, 2.188e-01, 1.971e-01, 1.477e-01, 7.030e-02, 3.858e-02, 2.411e-02, 1.719e-02, 1.321e-02, 1.004e-02]
ber_deepwave = [4.671e-01, 4.628e-01, 4.599e-01, 4.566e-01, 4.550e-01, 4.537e-01, 4.534e-01, 4.527e-01, 4.521e-01, 4.525e-01, 4.526e-01, 4.532e-01, 4.539e-01]
ber_mhsa = [2.638e-01, 2.465e-01, 2.278e-01, 2.094e-01, 1.857e-01, 1.180e-01, 3.790e-03, 9.326e-05, 1e-6, 1e-6, 1e-6, 1e-6, 1e-6]
ber_dat = [2.649e-01, 2.466e-01, 2.284e-01, 2.090e-01, 1.857e-01, 1.190e-01, 3.037e-03, 1e-6, 1e-6, 1e-6, 1e-6, 1e-6, 1e-6]

plt.figure(figsize=(9, 6), dpi=120)

plt.plot(snrs, ber_classical, label='Classical Baseline', color='black', marker='*', linestyle='-', linewidth=2, markersize=8, zorder=3)
plt.plot(snrs, ber_ours, label='Ours (Hybrid)', color='#2ecc71', marker='o', linestyle='-', linewidth=2.5, markersize=7, zorder=5)
plt.plot(snrs, ber_dat, label='DAT (Dual Attn)', color='#f39c12', marker='D', linestyle='-', linewidth=2, markersize=6, zorder=4)
plt.plot(snrs, ber_mhsa, label='Vanilla MHSA', color='#9b59b6', marker='^', linestyle=':', linewidth=2, markersize=7, zorder=4)
plt.plot(snrs, ber_deeprx, label='DeepRx', color='#3498db', marker='s', linestyle='--', linewidth=2, markersize=6, zorder=3)
plt.plot(snrs, ber_deepwave, label='DeepWaveform', color='#e74c3c', marker='X', linestyle='-.', linewidth=2, markersize=7, zorder=3)

plt.yscale('log')
plt.title("Post-FEC Bit Error Rate (BER) vs. $E_b/N_0$", fontsize=15, fontweight='bold', pad=15)
plt.xlabel(r"$E_b/N_0$ (dB)", fontsize=13)
plt.ylabel("Bit Error Rate (Log Scale)", fontsize=13)

plt.xlim(0, 12)
plt.ylim(1e-6, 1.0)

# Add a floor line to indicate zero errors
plt.axhline(y=1e-6, color='gray', linestyle='--', alpha=0.5)
plt.text(0.2, 1.2e-6, 'Zero Errors (Floor)', color='gray', fontsize=10)

plt.grid(True, which="major", ls="-", alpha=0.4)
plt.grid(True, which="minor", ls="--", alpha=0.2)
plt.legend(fontsize=11, loc='lower left')
plt.tight_layout()

plt.show()
```
