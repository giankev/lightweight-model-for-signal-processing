```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# ==============================================================================
# 2. LLR RELIABILITY DISTRIBUTION (AT Eb/N0 = 8 dB)
# ==============================================================================
# NOTE: Replace these with your actual flat numpy arrays of LLRs from the models

# Simulating highly confident LLRs for our physics-informed model (peaks far from 0)
llrs_ours = np.concatenate([np.random.normal(-4.5, 1.2, 5000), np.random.normal(4.5, 1.2, 5000)])

# Simulating less confident LLRs for a black-box model suffering from residual ICI
llrs_blackbox = np.concatenate([np.random.normal(-2.0, 2.5, 5000), np.random.normal(2.0, 2.5, 5000)])

plt.figure(figsize=(10, 6), dpi=120)

# Plot overlapping histograms
plt.hist(llrs_ours, bins=100, density=True, alpha=0.65, color='#2ecc71', label='Ours (Explicit Physics)', edgecolor='black', linewidth=0.5)
plt.hist(llrs_blackbox, bins=100, density=True, alpha=0.55, color='#e74c3c', label='Black-Box CNN (DeepRx/DeepWaveform)', edgecolor='black', linewidth=0.5)

# Formatting
plt.title("Empirical PDF of Log-Likelihood Ratios (LLRs) at $E_b/N_0 = 8$ dB", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("LLR Value (Log-Likelihood Ratio)", fontsize=13)
plt.ylabel("Probability Density", fontsize=13)

# Add decision boundary line
plt.axvline(x=0, color='black', linestyle='--', linewidth=1.5, alpha=0.8)
plt.text(0.5, 0.15, 'Decision Boundary\n(Highest Uncertainty)', rotation=90, va='center', fontsize=10, alpha=0.8)

plt.grid(True, axis='y', ls="--", alpha=0.5)
plt.legend(fontsize=11, loc='upper right')
plt.xlim(-10, 10)
plt.tight_layout()

plt.show()
```
