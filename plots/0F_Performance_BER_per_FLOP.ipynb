```python
import matplotlib.pyplot as plt
import numpy as np

# ==============================================================================
# 2. COMPUTATIONAL EFFICIENCY: RELIABILITY PER MFLOP
# ==============================================================================

# Models ordered by efficiency
labels = ['Ours (Hybrid)', 'DAT', 'Vanilla MHSA', 'DeepRx', 'DeepWaveform']
flops = np.array([72.57, 147.81, 162.30, 343.72, 295.99])
ber_6db = np.array([3.765e-02, 3.037e-03, 3.790e-03, 1.477e-01, 4.534e-01])

# Calculate Efficiency Metric: (1 - BER) / MFLOPs
# Represents "Successful decoding reliability per unit of compute"
efficiency_scores = (1 - ber_6db) / flops

# Colors
colors = ['#2ecc71', '#f39c12', '#9b59b6', '#3498db', '#e74c3c']

plt.figure(figsize=(10, 6), dpi=120)

# Create Bar Chart
bars = plt.bar(labels, efficiency_scores, color=colors, edgecolor='black', linewidth=1.2, width=0.6)

# Add values on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.0002, f'{yval:.4f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Formatting
plt.title("Hardware Efficiency: Decoding Reliability per MFLOP", fontsize=15, fontweight='bold', pad=15)
plt.ylabel(r"Efficiency Score $\left( \frac{1 - \text{BER}}{\text{MFLOPs}} \right)$", fontsize=13)
plt.xlabel("Receiver Architecture", fontsize=13)

# Add a subtle background highlight for the winning model
plt.axvspan(-0.5, 0.5, color='#2ecc71', alpha=0.1, zorder=0)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, max(efficiency_scores) * 1.15)
plt.tight_layout()

plt.show()
```
