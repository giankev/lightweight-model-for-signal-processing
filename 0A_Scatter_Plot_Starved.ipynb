```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# HARDCODED CONSTANTS FROM EVALUATION OUTPUT
# Models: Starved Condition (8 symbols / pilot)
# ==========================================

# X-axis: Eb/N0 range
ebno = np.arange(0, 13)

# 1. Classical Baseline BER (Same across all runs)
base_ber = [
    2.851e-01, 2.713e-01, 2.581e-01, 2.446e-01, 2.311e-01, 
    2.052e-01, 1.594e-01, 1.355e-01, 1.210e-01, 1.100e-01, 
    1.009e-01, 9.285e-02, 8.593e-02
]

# 2. Linear Attention SMART BER (Optimal Routing)
la_smart_ber = [
    2.670e-01, 2.502e-01, 2.336e-01, 2.163e-01, 1.951e-01, 
    1.477e-01, 2.846e-02, 9.071e-03, 7.418e-03, 5.051e-03, 
    2.538e-03, 1.503e-03, 5.229e-04
]

# 3. Squeeze-and-Excitation (SE) SMART BER
se_smart_ber = [
    2.709e-01, 2.543e-01, 2.365e-01, 2.196e-01, 1.984e-01, 
    1.520e-01, 3.011e-02, 1.299e-02, 1.099e-02, 9.310e-03, 
    8.399e-03, 7.829e-03, 8.271e-03
]

# 4. Gated Convolution (GLU) SMART BER
glu_smart_ber = [
    2.740e-01, 2.575e-01, 2.402e-01, 2.231e-01, 2.014e-01, 
    1.585e-01, 4.327e-02, 2.183e-02, 1.802e-02, 1.595e-02, 
    1.461e-02, 1.379e-02, 1.579e-02
]

# ==========================================
# PLOTTING LOGIC
# ==========================================

plt.figure(figsize=(10, 7))

# Plot Baseline
plt.semilogy(ebno, base_ber, marker='o', linestyle='--', 
             color='gray', linewidth=2, label='Classical Baseline')

# Plot Models
plt.semilogy(ebno, la_smart_ber, marker='s', linestyle='-', 
             color='#2ecc71', linewidth=2.5, markersize=7, label='Linear Attention')
plt.semilogy(ebno, se_smart_ber, marker='^', linestyle='-', 
             color='#3498db', linewidth=2.5, markersize=7, label='Squeeze & Excitation')
plt.semilogy(ebno, glu_smart_ber, marker='d', linestyle='-', 
             color='#e74c3c', linewidth=2.5, markersize=7, label='Gated Conv - GLU')

# Aesthetics
plt.title("Dynamic SNR-Aware Receiver Routing Performance\n(Starved Condition: 8 Data Symbols / Pilot)", 
          fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Eb/N0 (dB)", fontsize=13)
plt.ylabel("Post-FEC Bit Error Rate (BER)", fontsize=13)

# Grid & Limits
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.xlim(ebno[0], ebno[-1])

# Legend
plt.legend(fontsize=11, loc='lower left')
plt.tight_layout()

# Show plot
plt.show()
```
