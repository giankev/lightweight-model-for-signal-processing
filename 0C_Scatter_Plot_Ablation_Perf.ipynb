```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# DATA DEFINITIONS (BLER)
# ==========================================
ebno = np.arange(0, 13)

# 1. Standard Configuration (High Pilot Density)
# Original = Dynamic Aux Loss | Ablated = Static Aux Loss
std_orig_bler = [1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 9.980e-01, 9.205e-01, 3.850e-01, 2.850e-02, 8.000e-03, 7.000e-03, 7.000e-03, 7.500e-03]
std_abla_bler = [1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 9.990e-01, 9.170e-01, 3.825e-01, 3.050e-02, 9.000e-03, 9.000e-03, 9.000e-03, 9.500e-03]

# 2. Starved Configuration (Low Pilot Density)
# Original = Dynamic Aux Loss | Ablated = Static Aux Loss
stv_orig_bler = [1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 9.915e-01, 3.500e-01, 4.300e-02, 2.450e-02, 1.950e-02, 1.650e-02, 1.600e-02, 1.500e-02]
stv_abla_bler = [1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 1.000e+00, 9.930e-01, 4.440e-01, 4.900e-02, 1.800e-02, 1.450e-02, 1.350e-02, 1.200e-02, 1.050e-02]

# ==========================================
# PLOTTING
# ==========================================
plt.figure(figsize=(11, 7))

# Plot Standard (Non-Starved) - Cool Colors
plt.semilogy(ebno, std_orig_bler, marker='o', linestyle='-', color='#0984e3', linewidth=2.5, markersize=8, 
             label='Standard Original (Dynamic Aux)')
plt.semilogy(ebno, std_abla_bler, marker='s', linestyle='--', color='#00cec9', linewidth=2.5, markersize=8, 
             label='Standard Ablated (Static Aux)')

# Plot Starved - Warm Colors
plt.semilogy(ebno, stv_orig_bler, marker='^', linestyle='-', color='#e84393', linewidth=2.5, markersize=8, 
             label='Starved Original (Dynamic Aux)')
plt.semilogy(ebno, stv_abla_bler, marker='d', linestyle='--', color='#fdcb6e', linewidth=2.5, markersize=8, 
             label='Starved Ablated (Static Aux) - Winner')

# Formatting
plt.title("BLER Evolution: Standard vs Starved (Dynamic vs Static Aux Loss)", fontsize=15, fontweight='bold', pad=15)
plt.xlabel("Eb/N0 (dB)", fontsize=13)
plt.ylabel("Block Error Rate (BLER)", fontsize=13)
plt.grid(True, which="both", ls="--", alpha=0.6)
plt.xlim(0, 12)

# Set legend to lower left
plt.legend(fontsize=11, loc='lower left')

plt.tight_layout()
plt.show()
```

Results for starved models are expected to be worse. Starved models run on 8 symbols per pilot, while standard runs on 2 symbols per pilot.

The Starved model wins in the waterfall region because it has a better raw energy distribution for the data bits, and the neural network was forced to learn a much more robust, noise-resilient phase tracking algorithm instead of lazily relying on pilot interpolation.
